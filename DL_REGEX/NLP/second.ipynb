```python
# TF(term frequency)
# IDF(Inverse document frequencey)
```

```python
#TF(t,d) = (Number of occurence of term t in document d)/
#(total number of term in the document d)
#ex = 

#d1 = people watch cricket.

#tf(people,d1) = 1/3

#d2 = match was match
#tf(match,d2) = 2/3
```

```python
# IDF(t) = loge(total number of  documnet in the corpus)/
#(number of document with term t in them)
#ex == 

#d1 = people watch cricket.
#d2 = cricket was cricket.
#d3 = people write comment.
#d4 = cricket write comment.

#idf(cricket) = Log(4/4)
#idf(people) = Log(4/2)
```
