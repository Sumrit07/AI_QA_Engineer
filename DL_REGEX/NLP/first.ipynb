step2 remove html

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
text = "<html><body> Sumrit<p> Click <a href = .com</a>"
```

```python
remove_html_tags(text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

```python
def remove_url(text):
    pattern = re.compile(r'https?://S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
text4 = 'For click https://www.deture.com/delta/note12234acc'
```

```python
df['review'] = df['review'].apply(remove_url)
```

step -5 chatword

```python
chat_words  = {'u2':'you too','BBL':'Be Back Later'}
```

```python
chat_words
```

```python
def chat_conversation(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
            
    return "".join(new_text)
```

```python
chat_conversation('BBL')
```

step 7 removing stepwords

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
STOPWORDS = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word in set (stopwords.words('english')):
            new_text.append('')
        else:
            new_text.append(word)
            
    x  = new_text[:]
    new_text.clear()
    return "".join(x)


remove_stopwords('pro my all-time move, a stroy of self')
```

Stemming

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word)for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```
