```python
from googletrans import Translator

translator = Translator()

hinglish_text = "1. The overall experience was really good and everything worked smoothly.\n" "2. Mujhe interface kaafi clean aur easy to use laga, isliye use karna simple tha.\n" "3. Performance fast thi aur kisi major issue ka saamna nahi hua.\n" "4. Features useful hain aur daily use ke liye kaafi practical lage.\n" "5. Design modern aur attractive hai, jo overall experience ko better banata hai.\n" "6. Customer support bhi responsive tha aur queries jaldi resolve ho gayi.\n" "7. Thode bahut improvements ki zarurat hai, lekin overall quality impressive hai.\n" "8. Price ke hisaab se product/service achhi value provide karta hai.\n"  "9. Main is experience se satisfied hoon aur future mein bhi use karna pasand karunga.\n" "10. Overall, I would definitely recommend it to others because it is reliable, user-friendly, and worth trying."

translated = await translator.translate(hinglish_text, dest="en")

print("Original:\n",hinglish_text)
print()
print("---------------------------------------")
print()
print("English:\n",translated.text)
```

```python
from deep_translator import GoogleTranslator
from textblob import TextBlob
import re

# Hinglish paragraph
hinglish_text = """
Yeh phone ka camera bahut accha hai. Battery backup bhi kaafi achha chal raha hai.
Lekin charging speed thodi slow hai. Display bahut smooth lagti hai.
Overall mujhe yeh phone pasand aaya.
"""

# Hinglish -> English
english_text = GoogleTranslator(
    source="auto",
    target="en"
).translate(hinglish_text)

print("Translated Paragraph:\n")
print(english_text)

# Sentence Splitting
sentences = re.split(r'(?<=[.!?])\s+', english_text)

print("\nSentence-wise Sentiment:\n")

for i, sentence in enumerate(sentences, 1):
    if sentence.strip():
        blob = TextBlob(sentence)
        polarity = blob.sentiment.polarity

        if polarity > 0:
            sentiment = "Positive "
        elif polarity < 0:
            sentiment = "Negative "
        else:
            sentiment = "Neutral "

        print(f"Sentence {i}: {sentence}")
        print(f"Polarity : {polarity:.3f}")
        print(f"Sentiment: {sentiment}")
        print("-"*50)
```
